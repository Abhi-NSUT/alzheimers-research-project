```python
# Cell 1: Required imports
import os
import random
import tensorflow as tf
from tensorflow import keras
import numpy as np
import pandas as pd
import pickle  # Standard pickle (no need for pickle5 if using Python 3.8+)
import matplotlib.pyplot as plt
from keras.models import Sequential
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import classification_report
from keras.layers import Dense, Dropout, MaxPooling2D, Flatten, Conv2D

```

```python
# Cell 2: Function to reset random seeds for reproducibility
def reset_random_seeds(seed):
    os.environ['PYTHONHASHSEED'] = str(seed)
    tf.random.set_seed(seed)
    np.random.seed(seed)
    random.seed(seed)
```

```python
# Cell 3: Load the data
with open("img_train.pkl", "rb") as fh:
    data = pickle.load(fh)
X_train_ = pd.DataFrame(data)["img_array"]

with open("img_test.pkl", "rb") as fh:
    data = pickle.load(fh)
X_test_ = pd.DataFrame(data)["img_array"]

with open("img_y_train.pkl", "rb") as fh:
    data = pickle.load(fh)
y_train = np.array(pd.DataFrame(data)["label"].values.astype(np.float32)).flatten()

with open("img_y_test.pkl", "rb") as fh:
    data = pickle.load(fh)
y_test = np.array(pd.DataFrame(data)["label"].values.astype(np.float32)).flatten()
```

```python
# Cell 4: Adjust label values (if required)
y_test[y_test == 2] = -1
y_test[y_test == 1] = 2
y_test[y_test == -1] = 1

y_train[y_train == 2] = -1
y_train[y_train == 1] = 2
y_train[y_train == -1] = 1
```

```python
# Cell 5: Convert to arrays
X_train = np.array([X_train_.values[i] for i in range(len(X_train_))])
X_test = np.array([X_test_.values[i] for i in range(len(X_test_))])
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Conv2D, MaxPooling2D, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.applications import MobileNetV2
from sklearn.metrics import classification_report
import numpy as np

# Add lists to store metrics for each model
model_accuracies = {}
model_precisions = {}
model_recalls = {}
model_f1_scores = {}

# Define different models in a dictionary
models = {
    "Simple_CNN": Sequential([
        Conv2D(100, (3, 3), activation='relu', input_shape=(72, 72, 3)),
        MaxPooling2D((2, 2)),
        Dropout(0.5),
        Conv2D(50, (3, 3), activation='relu'),
        MaxPooling2D((2, 2)),
        Dropout(0.3),
        Flatten(),
        Dense(3, activation="softmax")
    ]),
    
    "Deeper_CNN": Sequential([
        Conv2D(64, (3, 3), activation='relu', input_shape=(72, 72, 3)),
        MaxPooling2D((2, 2)),
        Conv2D(128, (3, 3), activation='relu'),
        MaxPooling2D((2, 2)),
        Conv2D(256, (3, 3), activation='relu'),
        MaxPooling2D((2, 2)),
        Flatten(),
        Dense(128, activation='relu'),
        Dropout(0.5),
        Dense(3, activation="softmax")
    ]),

    "MobileNetV2": Sequential([
        MobileNetV2(input_shape=(72, 72, 3), include_top=False, weights='imagenet', pooling='avg'),
        Dense(3, activation="softmax")
    ])
}

# Iterate over seeds and models
for seed in seeds:
    reset_random_seeds(seed)
    
    for model_name, model in models.items():
        # Compile the model
        model.compile(Adam(learning_rate=0.001), "sparse_categorical_crossentropy", metrics=["sparse_categorical_accuracy"])
        
        # Print model summary
        print(f"Training {model_name} with seed {seed}")
        model.summary()

        # Train the model
        history = model.fit(X_train, y_train, epochs=50, batch_size=32, validation_split=0.1, verbose=1)
        
        # Evaluate the model
        score = model.evaluate(X_test, y_test, verbose=0)
        print(f'{model_name} - Test loss: {score[0]} / Test accuracy: {score[1]}')
        
        # Initialize metric lists for each model if not already
        if model_name not in model_accuracies:
            model_accuracies[model_name] = []
            model_precisions[model_name] = []
            model_recalls[model_name] = []
            model_f1_scores[model_name] = []
        
        # Append accuracy
        model_accuracies[model_name].append(score[1])

        # Make predictions on test data
        test_predictions = model.predict(X_test)
        test_label = np.argmax(to_categorical(y_test, 3), axis=1)
        predicted_label = np.argmax(test_predictions, axis=1)

        # Generate classification report and append performance metrics
        cr = classification_report(test_label, predicted_label, output_dict=True)
        model_precisions[model_name].append(cr["macro avg"]["precision"])
        model_recalls[model_name].append(cr["macro avg"]["recall"])
        model_f1_scores[model_name].append(cr["macro avg"]["f1-score"])
```

```python
import numpy as np

# Calculate and display the average accuracy for each model
for model_name, accuracies in model_accuracies.items():
    avg_accuracy = np.mean(accuracies)
    print(f"Average accuracy for {model_name}: {avg_accuracy:.4f}")
```

```python
import matplotlib.pyplot as plt

# Function to plot the performance metrics
def plot_metrics(metric_dict, metric_name):
    plt.figure(figsize=(10, 6))
    
    for model_name, metric_values in metric_dict.items():
        plt.plot(metric_values, label=model_name, marker='o')  # Line plot with markers
    
    plt.title(f'{metric_name} Across Different Seeds')
    plt.xlabel('Seed Index')
    plt.ylabel(metric_name)
    plt.xticks(ticks=range(len(list(metric_dict.values())[0])), labels=[f'Seed {i+1}' for i in range(len(list(metric_dict.values())[0]))])
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend()
    plt.show()


# Plot Accuracy for each model
plot_metrics(model_accuracies, 'Accuracy')

# Plot Precision for each model
plot_metrics(model_precisions, 'Precision')

# Plot Recall for each model
plot_metrics(model_recalls, 'Recall')

# Plot F1 Score for each model
plot_metrics(model_f1_scores, 'F1 Score')
```
